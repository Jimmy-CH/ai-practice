```python
import pandas as pd


FILE = '智能家居环境控制系统数据集.xlsx'
df = pd.read_excel(FILE)

COL_TIME = '时间戳'
COL_TEMP = '温度'
COL_HUM = '湿度'
COL_LIGHT = '光照水平'
COL_RESP = '响应时间'
COL_ENERGY = '能源消耗'

print('数据规模:', df.shape)
print('列名:', list(df.columns))
print('\n前5行:')
print(df.head())
print('\n描述统计:')
print(df.describe().round(2))
```

```python
df[COL_TIME] = pd.to_datetime(df[COL_TIME])
df['小时'] = df[COL_TIME].dt.hour

print('时间范围:', df[COL_TIME].min(), '~', df[COL_TIME].max())
print('总记录数:', len(df))
print('缺失值:', df.isnull().sum().sum())
print('新数据:')
print(df.head())
```

```python
# 1.1 各小时环境参数均值
hourly_pref = df.groupby('小时').agg(
    温度均值=(COL_TEMP, 'mean'),
    湿度均值=(COL_HUM, 'mean'),
    光照均值=(COL_LIGHT, 'mean'),
    记录数=(COL_TEMP, 'count')
).round(2)
print('=== 各小时环境偏好 ===')
print(hourly_pref)

temp_peak_h = hourly_pref['温度均值'].idxmax()
temp_low_h = hourly_pref['温度均值'].idxmin()
hum_peak_h = hourly_pref['湿度均值'].idxmax()
light_peak_h = hourly_pref['光照均值'].idxmax()
light_low_h = hourly_pref['光照均值'].idxmin()
print(f'\n温度最高: {temp_peak_h}点 ({hourly_pref.loc[temp_peak_h, "温度均值"]}°C)')
print(f'温度最低: {temp_low_h}点 ({hourly_pref.loc[temp_low_h, "温度均值"]}°C)')
print(f'湿度最高: {hum_peak_h}点 ({hourly_pref.loc[hum_peak_h, "湿度均值"]}%)')
print(f'光照最高: {light_peak_h}点 ({hourly_pref.loc[light_peak_h, "光照均值"]} lux)')
print(f'光照最低: {light_low_h}点 ({hourly_pref.loc[light_low_h, "光照均值"]} lux)')
```

```python
# 1.2 时段划分汇总
def day_period(h):
    if 6 <= h < 12:
        return '上午(6-12)'
    if 12 <= h < 18:
        return '下午(12-18)'
    if 18 <= h < 22:
        return '傍晚(18-22)'
    return '夜间(0-6/22-24)'

df['时段'] = df['小时'].apply(day_period)
period_pref = df.groupby('时段')[[COL_TEMP, COL_HUM, COL_LIGHT]].mean().round(2)
period_pref = period_pref.reindex(['上午(6-12)', '下午(12-18)', '傍晚(18-22)', '夜间(0-6/22-24)'])
print('=== 各时段环境偏好均值 ===')
print(period_pref)
```

```python
resp_stats = df[COL_RESP].agg(['mean', 'median', 'std', 'min', 'max']).round(3)
print('=== 响应时间总体统计 ===')
print(resp_stats)
print(f'\n全库平均响应时间: {df[COL_RESP].mean():.3f} 秒')

SLOW_THRESHOLD = 4.0
slow_count = (df[COL_RESP] > SLOW_THRESHOLD).sum()
print(f'慢响应（>{SLOW_THRESHOLD}s）: {slow_count} 条，占比 {slow_count/len(df)*100:.1f}%')
```

```python
# 2.1 各小时响应时间
hourly_resp = df.groupby('小时')[COL_RESP].agg(['mean', 'count', 'max']).round(3)
hourly_resp.columns = ['平均响应', '记录数', '最大响应']
hourly_resp['影响指数'] = (hourly_resp['平均响应'] * hourly_resp['记录数']).round(1)
print('=== 各小时响应时间 ===')
print(hourly_resp.sort_values('平均响应', ascending=False).head(8))

slowest_hours = hourly_resp['平均响应'].nlargest(3)
fastest_hours = hourly_resp['平均响应'].nsmallest(3)
bottleneck_hour = hourly_resp['影响指数'].idxmax()
print(f'\n响应最慢时段: {slowest_hours.index.tolist()} 点')
print(f'响应最快时段: {fastest_hours.index.tolist()} 点')
print(f'综合影响最大时段: {bottleneck_hour} 点')
```

```python
# 2.2 响应时间与各环境参数的相关性
corr_resp = {c: df[COL_RESP].corr(df[c]) for c in [COL_TEMP, COL_HUM, COL_LIGHT, COL_ENERGY]}
corr_series = pd.Series(corr_resp).round(3)
print('=== 响应时间与环境/能耗相关性 ===')
print(corr_series)

period_resp = df.groupby('时段')[COL_RESP].mean().round(3)
period_resp = period_resp.reindex(['上午(6-12)', '下午(12-18)', '傍晚(18-22)', '夜间(0-6/22-24)'])
print('\n=== 各时段平均响应 ===')
print(period_resp)
slowest_period = period_resp.idxmax()
```

```python
energy_stats = df[COL_ENERGY].agg(['mean', 'median', 'std', 'min', 'max']).round(3)
print('=== 能源消耗总体统计 ===')
print(energy_stats)
print(f'\n全库平均能耗: {df[COL_ENERGY].mean():.3f}')

HIGH_Q = df[COL_ENERGY].quantile(0.75)
LOW_Q = df[COL_ENERGY].quantile(0.25)
high_energy = df[df[COL_ENERGY] >= HIGH_Q]
low_energy = df[df[COL_ENERGY] <= LOW_Q]
print(f'高能耗阈值（75分位）: {HIGH_Q:.2f}，共 {len(high_energy)} 条')
print(f'低能耗阈值（25分位）: {LOW_Q:.2f}，共 {len(low_energy)} 条')
```

```python
# 3.1 各小时/时段能耗
hourly_energy = df.groupby('小时')[COL_ENERGY].agg(['mean', 'count']).round(3)
hourly_energy.columns = ['平均能耗', '记录数']
print('=== 能耗最高的小时 Top5 ===')
print(hourly_energy.sort_values('平均能耗', ascending=False).head(5))
print('\n=== 能耗最低的小时 Top5 ===')
print(hourly_energy.sort_values('平均能耗').head(5))

period_energy = df.groupby('时段')[COL_ENERGY].mean().round(3)
period_energy = period_energy.reindex(['上午(6-12)', '下午(12-18)', '傍晚(18-22)', '夜间(0-6/22-24)'])
print('\n=== 各时段平均能耗 ===')
print(period_energy)
high_energy_period = period_energy.idxmax()
low_energy_period = period_energy.idxmin()
high_energy_hours = hourly_energy['平均能耗'].nlargest(3).index.tolist()
low_energy_hours = hourly_energy['平均能耗'].nsmallest(3).index.tolist()
```

```python
# 3.2 高/低能耗场景特征对比（节能潜力）
compare_cols = [COL_TEMP, COL_HUM, COL_LIGHT, COL_RESP]
high_profile = high_energy[compare_cols].mean().round(2)
low_profile = low_energy[compare_cols].mean().round(2)
energy_compare = pd.DataFrame({'高能耗组': high_profile, '低能耗组': low_profile})
print('=== 高/低能耗组环境特征对比 ===')
print(energy_compare)

saving_potential = (df[COL_ENERGY].mean() - low_energy[COL_ENERGY].mean()) / df[COL_ENERGY].mean() * 100
print(f'\n若高能耗时段降至低能耗水平，理论节能潜力约 {saving_potential:.1f}%')
```
